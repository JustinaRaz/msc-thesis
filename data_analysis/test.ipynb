```python
def compute_basic_stats(folder_path):
    total_tokens = 0
    total_sentences = 0
    total_characters = 0
    types = set()

    current_sentence_length = 0

    for file in Path(folder_path).glob("*.pos"):
        with open(file, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()

                # --- Sentence boundary ---
                if not line:
                    if current_sentence_length > 0:
                        total_sentences += 1
                        current_sentence_length = 0
                    continue

                parts = line.split("\t")
                if len(parts) < 2:
                    continue

                token = parts[0]

                # --- Update counts ---
                total_tokens += 1
                current_sentence_length += 1
                total_characters += len(token)
                types.add(token.lower())  # lowercase for TTR

        # catch last sentence if file doesn't end with newline
        if current_sentence_length > 0:
            total_sentences += 1
            current_sentence_length = 0

    # --- Metrics ---
    avg_sentence_length = (
        total_tokens / total_sentences if total_sentences else 0
    )

    avg_word_length = (
        total_characters / total_tokens if total_tokens else 0
    )

    ttr = (
        len(types) / total_tokens if total_tokens else 0
    )

    return {
    "tokens": total_tokens,
    "sentences": total_sentences,
    "types": len(types),
    "avg_sentence_length": round(avg_sentence_length, 1),
    "avg_word_length": round(avg_word_length, 1),
    "ttr": round(ttr, 3),
}

results = {}
base_path = Path("data_analysis/data/morphological_annotations/pedagogic_lithuanian_corpus")

for level in ["a1", "b1", "c1"]:
    folder = base_path / level
    results[level] = compute_basic_stats(folder)

for level, res in results.items():
    print(level, res)
```
